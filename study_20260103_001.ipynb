map() 함수를 사용해 하나로 묶어 보자

```python
W_R = input().split()   # W_R를 뛰어서 받아라 W는 0리스트에 들어가고 R은 1리스트레 들어가라
W = int(W_R[0])
R = int(W_R[1])

one_rm = W * (1 + R / 30) # W와 R을 계산식에 계산하여 one_rm의 값을 구한다.
print(int(one_rm))  #one_rm을 정수로 표현하여 출력한다.
```

```python
# map() 함수를 사용해

W,R = map(int,input().split())

oen_rm = W * ( 1 + R / 30 )

print(int(oen_rm))
```
