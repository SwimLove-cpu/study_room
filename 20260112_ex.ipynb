```python
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.application import MIMEApplication
from getpass import getpass
import os

def send_email():
    """대화형 이메일 발송 프로그램"""

    print("=" * 50)
    print("이메일 발송 프로그램")
    print("=" * 50)

    # 발신자 정보 입력
    sendEmail = input("발신자 이메일 주소: ")
    password = getpass("발신자 이메일 비밀번호(앱 비밀번호): ")

    # 수신자 정보 입력
    recvEmail = input("수신자 이메일 주소: ")

    # 메일 제목과 내용 입력
    subject = input("메일 제목: ")
    print("메일 내용을 입력하세요 (Enter 두 번으로 종료):")

    content_lines = []
    empty_count = 0
    while True:
        line = input()
        if line == "":
            empty_count += 1
            if empty_count >= 2:
                break
        else:
            empty_count = 0
            content_lines.append(line)

    content = "\n".join(content_lines)

    # 첨부파일 여부 확인
    attach_file = input("첨부파일이 있습니까? (y/n): ").lower()

    # SMTP 서버 설정
    smtpName = "smtp.gmail.com"
    smtpPort = 587

    try:
        # 메시지 생성
        msg = MIMEMultipart()
        msg['Subject'] = subject
        msg['From'] = sendEmail
        msg['To'] = recvEmail

        # 본문 추가
        text = MIMEText(content, _charset="utf-8")
        msg.attach(text)

        # 첨부파일 처리
        if attach_file == 'y':
            filename = input("첨부할 파일 경로: ")
            if os.path.exists(filename):
                with open(filename, 'rb') as etcFD:
                    etcPart = MIMEApplication(etcFD.read())
                    # 첨부파일 헤더 추가
                    basename = os.path.basename(filename)
                    etcPart.add_header('Content-Disposition', 'attachment', filename=basename)
                    msg.attach(etcPart)
                print(f"'{basename}' 파일이 첨부되었습니다.")
            else:
                print("파일을 찾을 수 없습니다. 첨부파일 없이 진행합니다.")

        # SMTP 서버 연결 및 메일 발송
        print("\n메일 발송 중...")
        s = smtplib.SMTP(smtpName, smtpPort)
        s.starttls()
        s.login(sendEmail, password)
        s.sendmail(sendEmail, recvEmail, msg.as_string())
        s.close()

        print("\n✅ 메일이 성공적으로 발송되었습니다!")

    except smtplib.SMTPAuthenticationError:
        print("\n❌ 인증 실패: 이메일 주소나 비밀번호를 확인하세요.")
    except Exception as e:
        print(f"\n❌ 메일 발송 실패: {e}")

if __name__ == "__main__":
    send_email()
```
